```python
import torch

x = torch.arange(16)
print(x)
print(x.shape)
```

```python
reshaped_x = x.reshape(4, 4)
print(reshaped_x)
print(reshaped_x.shape)
```

```python
tensor = torch.tensor([[[1,2,3,4]]])
print(tensor)
print(tensor.shape)
```

```python
squeezed_tensor = tensor.squeeze()
print(squeezed_tensor)
print(squeezed_tensor.shape)
```

```python
squeezed_dim_tensor = tensor.squeeze(0)
print(squeezed_dim_tensor)
print(squeezed_dim_tensor.shape)
```
