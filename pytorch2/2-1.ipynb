Sequential을 사용하여 신경망 모델 만들기

```python
model = nn.Sequential(
    nn.Linear(784, 256),
    nn.ReLU(),
    nn.Dropout(0.5),
    nn.Linear(256, 128),
    nn.ReLU(),
    nn.Dropout(0.5),
    nn.Linear(128, 10)
)

print(model)
```

Module 상속 클래스로 모델 만들기

```python
class CustomModel(nn.Module):
    def __init__(self):
        super(CustomModel, self).__init__()

        self.fc1 = nn.Linear(784, 128)
        self.relu1 = nn.ReLU()

        self.fc2 = nn.Linear(128, 64)
        self.relu2 = nn.ReLU()

        self.fc3 = nn.Linear(64, 10)

        self.dropout = nn.Dropout(0.2)

def forward(self, x):
    x = self.relu1

    x = self.dropout

    x = self.relu2(self.fc2(x))

    x = self.fc3(x)

    return nn.Softmax(dim=1)(x)

model = CustomModel()
model
```
