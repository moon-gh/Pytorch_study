𝑦=𝑊𝑥+𝑏

```python
import torch

def custom_layer(x, weight, bias):
    x = torch.matmul(weight, x) + bias
    return x
```

```python
input_tensor = torch.randn(10, 1)

weight = torch.randn(2, 10)
bias = torch.randn(2)

output = custom_layer(input_tensor, weight, bias)
print(output)
```

𝑦=𝑠𝑖𝑛(𝑥𝑊𝑇)+𝑏

```python
def custom_sin_layer(x, weight, bias):
    x = torch.matmul(x, weight.t())
    x = torch.sin(x)
    return x
```

```python
input_tensor = torch.randn(1, 10)

weight = torch.randn(2,10)
bias = torch.randn(1)

output = custom_sin_layer(input_tensor, weight, bias)
output
```

```python
import torch.nn as nn

class CustomLinearLayer(nn.Module):
    def __init__(self, input_dim, output_dim):
        super(CustomLinearLayer, self).__init__()

        self.weight = nn.Parameter(torch.randn(output_dim, input_dim))
        self.bias = nn.Parameter(torch.randn(output_dim))

    def forward(self, x):
        y = torch.matmul(x, self.weight.t()) + self.bias
        return y
```

```python
import torch.nn as nn

class CustomLinearLayer(nn.Module):
    def __init__(self, input_dim, output_dim):
        super(CustomLinearLayer, self).__init__()

        self.weight = nn.Parameter(torch.randn(output_dim, input_dim))
        self.bias = nn.Parameter(torch.randn(output_dim))

    def forward(self, x):
        y = torch.matmul(x, self.weight.t()) + self.bias
        return y
```

```python
input_dim = 5
output_dim = 3

custom_layer = CustomLinearLayer(input_dim, output_dim)

x = torch.randn(2, input_dim)

y = custom_layer(x)

print(y)
```

CustomLinearLayer를 이용해 딥러닝 모델을 정의

```python
import torch.nn.functional as F

class CustomMLP(nn.Module):
    def __init__(self):
        super(CustomMLP, self).__init__()
        self.fc1 = CustomLinearLayer(20, 50)
        self.fc2 = CustomLinearLayer(50,30)
        self.fc3 = CustomLinearLayer(30, 3)

    def forward(self, x):
        x = F.relu(self.fc1(x))
        x = F.relu(self.fc2(x))
        x = self.fc3(x)
        return x
```

```python
model = CustomMLP()

x = torch.randn(5, 20)

output = model(x)

print(output)
```
